Create a class BankAccount with __init__ setting self.balance = starting_balance. Add a method deposit(self, amount)
that adds amount to self.balance, and a method withdraw(self, amount) that subtracts amount only if there is enough
balance, otherwise it should return the string "Insufficient funds".

```python
class BankAccount:
    def __init__(self,owner,balance):
        self.owner= owner
        self.balance=balance
    def deposit(self,amount):
        self.balance = self.balance + amount
    def withdraw(self,amount):
        if amount>self.balance:
            return "Insufficient funds"
        else:
            self.balance=self.balance-amount
neesha=BankAccount("Neesha",10000)
print(neesha.owner,neesha.balance)
neesha.deposit(2000)
print(neesha.balance)
neesha.withdraw(14000)


```

Create a parent class Employee with __init__ setting self.name and self.base_salary, and a method salary(self) that
returns self.base_salary. Create a child class Manager(Employee) that overrides salary(self) to return self.base_salary
+ 20000 (a fixed bonus).

```python
class Employee:
    def __init__(self, name, base_salary):
        self.name= name
        self.base_salary= base_salary
    def salary(self):
        return self.base_salary
class Manager(Employee):
    def salary(self):
        return self.base_salary + 2000 #(fixed bonus)
e= Employee("Redha", 90000)
m= Manager("Neesha", 100000)
print(e.salary())
print(m.salary())

```
