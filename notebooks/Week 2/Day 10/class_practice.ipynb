that adds amount to self.balance, and a method withdraw(self, amount) that subtracts amount only if there is enough
balance, otherwise it should return the string "Insufficient funds".

```python
class BankAccount:
    def __init__(self,owner,balance):
        self.owner= owner
        self.balance=balance
    def deposit(self,amount):
        self.balance = self.balance + amount
    def withdraw(self,amount):
        if amount>self.balance:
            return "Insufficient funds"
        else:
            self.balance=self.balance-amount
neesha=BankAccount("Neesha",10000)
print(neesha.owner,neesha.balance)
neesha.deposit(2000)
print(neesha.balance)
neesha.withdraw(14000)


```

### Build a small hierarchy: a parent class Shape with __init__ setting self.name, and a method area(self) that returns 0.
### Create child classes Square(Shape) and Circle(Shape), each overriding area() using a side or radius attribute you add in
### their own __init__.

```python
class Shape:
    def __init__(self,name):
        self.name=name
    def area(self):
        return 0
    def __str__(self):
        return f"{self.name}  has area: {self.area()}"
class Square(Shape):
    def __init__(self,name,l,w):
        super().__init__(name)
        self.l=l
        self.w=w
    def area(self):
        result =self.l*self.w
        return result
class Circle(Shape):
    def __init__(self,name,r):
        super().__init__(name)
        self.r=r
    def area(self):
        result= self.r**2 * 3.14
        return result
sq= Square("Square",2,3)
cl= Circle("Circle",2)
print(sq)
print(cl)
```
