```python
import pandas as pd
df = pd.DataFrame({
"name": ["Ali", "Sara", "Bilal"],
"city": ["Karachi", "Lahore", "Karachi"],
"score": [85, 92, 78],
})
print(df["name"]) # Series of names
print(df[["name", "score"]]) # DataFrame, 2 columns
print(df.loc[1]) # Sara's full row, by label
print(df.iloc[0])
```

```python
import pandas as pd
df = pd.DataFrame({
"name": ["Ali", "Sara", "Bilal", "Hina"],
"city": ["Karachi", "Lahore", "Karachi", "Islamabad"],
"score": [85, 92, 78, 88],
})
high_scorers = df[df["score"] > 80]
print(high_scorers) # Ali, Sara, Hina
karachi_high = df[(df["city"] == "Karachi") & (df["score"] > 80)]
print(karachi_high) # Ali only
```
